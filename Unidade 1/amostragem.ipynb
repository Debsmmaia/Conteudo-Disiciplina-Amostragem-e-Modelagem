Amostragem Aleatória Simples 

Criamos uma lista com valores aleatórios entre 0 e 1 com a mesma quantidade de valores do DF, depois colocamos esses "rotulos" em cada uma das linhas do DF, e depois com base no 0 ou 1 separamos em "caixas"

```python
import pandas as pd 
import numpy as np 
```

```python
dados = pd.read_csv("dados/iris.csv")
dados.shape #ver a dimensão da base de dados
np.random.seed(9999) #define a semente aleatória para repetir sempre os mesmos dados aleatórios
```

```python
amostra = np.random.choice(a=[0,1], size=150, replace=True,p=[0.7,0.3])
```

Amostragem Sistemática

Calculamos o valor de k com base na população e na amostra, sendo k = populacao / amostra

Depois escolhemos um valor ateatório entre 1 e k para ser o ponto de partida

Depois criamos um loop para somar o valor inicial mais k pela quantidade da amostra = 15

Após criar os rotulos aleatórios, escolhemos dentro do DF os valores correspondentes as posições criadas aleatoriamente 


```python
populacao = 150
amostra = 15
k = populacao / amostra #Fator sistematico - o quanto "salta"
print(k)
```

```python
r = np.random.randint(low=1, high=k+1, size=1)
#Valor aleatório para iniciar a amostra, iniciando de 1 até o k+1 (de 1 até 11)
print(r)
```

```python
acumulador = r[0] #valor para iniciar a busca
sorteados = [] #lista vazia para adicionar os valores
for i in range(amostra):
    sorteados.append(acumulador) #adiciona na lista dos sorteados o valor inicial somado ao K(10)
    acumulador += k
print([int(x) for x in sorteados])
```

```python
amostra_sistematica = dados.loc[sorteados]
amostra_sistematica
```
