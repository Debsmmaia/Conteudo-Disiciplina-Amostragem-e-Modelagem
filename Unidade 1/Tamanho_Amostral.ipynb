```python
import math
import matplotlib.pyplot as plt
import numpy as np
from scipy.stats import norm, t
```

Encontrando N - Pela Normal (Z)

```python
nivel_confianca = 0.95
margem_erro = 50
sigma = 400
```

```python
z_score = norm.ppf((1+ nivel_confianca) / 2) #encontra valor de Z pela normal
n = (z_score * (sigma/margem_erro)) ** 2 #calcula o valor da amostra
tamanho_amostra = math.ceil(n) #arredonda o valor de N para numero inteiro

print("Tamanho da amostra: ", tamanho_amostra)
```

```python
nivel_confianca = 0.995
margem_erro = 1.5
variancia = 23
sigma = math.sqrt(variancia)

z_score = norm.ppf((1 + nivel_confianca) / 2)
n = (z_score * sigma / margem_erro) ** 2
tamanho = math.ceil(n)
tamanho
```

Encontrando N - Para populações com tamanho conhecido

Foi feito de duas formas (geram o mesmo resultado):

1° - Fórmula com correção

2° - Fórmula direta 

```python
# UTILIZANDO AJUSTE NO VALOR DE N   

nivel_confianca = 0.92
margem_erro = 50
tamanho_pop = 386

z_score = norm.ppf((1 + nivel_confianca) / 2)
n = (z_score**2 * sigma**2) / (margem_erro**2) #calculo normal, como se a pop fosse infinita
n_ajustado = n / (1 + ((n-1) / tamanho_pop)) #correção do valor de N para pop finita

tamanho_amostra = math.ceil(n_ajustado)
tamanho_amostra
```

```python
# UTILIZANDO A FORMULA COMPLETA EM UMA ÚNICA LINHA
n = ((z_score**2) * (sigma**2) * tamanho_pop) / ((z_score**2) * (sigma**2) + (tamanho_pop - 1) * (margem_erro**2))

tamanho_amostra = math.ceil(n)
tamanho_amostra
```

Encontrando N - Variância desconhecida (T-Student)

```python
nivel_confianca = 0.95
erro = 60
s = 420  # desvio padrão da amostra
n_aproximado = 25  # estimativa inicial para graus de liberdade

# Valor crítico t para 95% de confiança (bicaudal), com 24 graus de liberdade (n - 1)
t_valor = t.ppf((1 + nivel_confianca) / 2, df=n_aproximado - 1)

# Fórmula do tamanho da amostra: n = (t * s / E)²
n_t = (t_valor * s / erro) ** 2

n_t = math.ceil(n_t)

n_t
```

Encontrando N - Para proporção

```python
nivel_confianca = 0.99
e = 0.05
p = 0.23 #proporção estimada (se n tiver uma estimativa anterior, usamos 0.5)

z = norm.ppf((1 + nivel_confianca) / 2)
n_p = (z**2 * p * (1 - p) / e**2)

n_p = math.ceil(n_p)
n_p
```

```python
p_conservador = 0.5 #usando proporção prévia conservadora
n_conservador = (z**2 * p_conservador * (1 - p_conservador)) / e**2
n_conservador = math.ceil(n_conservador)
n_conservador
```

```python
p = 0.10
nivel_confianca = 0.92
e = 0.02

z = norm.ppf((1 + nivel_confianca) / 2)
n_p = (z**2 * p * (1 - p) / e**2)

n_p = math.ceil(n_p)
n_p
```

Correção - Quando a população é conhecida para a Proporção

```python
tamanho_pop = 550

# n0 = tamanho de amostra inicial, calculado antes com p = 0.5 (caso conservador),
# ou seja, o tamanho que você precisaria se a população fosse "infinita"
n0 = n_conservador

# Correção para população finita: n = (n0 * N) / (n0 + N - 1)
n_ajustada = (n0 * tamanho_pop) / (n0 + tamanho_pop - 1)
n_ajustada = math.ceil(n_ajustada)

n_ajustada 
```
