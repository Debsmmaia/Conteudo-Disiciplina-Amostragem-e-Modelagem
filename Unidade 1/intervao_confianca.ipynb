```python
import pandas as pd
import numpy as np
import scipy.stats as stats
import matplotlib.pyplot as plt
```

```python
media = 346
sigma = 3
n = 120
alpha = 0.5
```

```python
zc = stats.norm.ppf(1 - alpha/2)
zc
```

```python
erro = zc *sigma/np.sqrt(n)
erro
```

```python
intervalo_confianca = (media - erro, media + erro)
intervalo_confianca
```
